```python
"""Load the residual load in enappsys."""

import datetime as dt

from zoneinfo import ZoneInfo

from impuls_algo.interfaces.enappsys import EnappsysInterface, EnappsysParameter
from impuls_algo.interfaces.settings import EnappsysSettings
from impuls_algo.models.data import EnappsysDataId
from impuls_interfaces.aws import BaseSsmParameters
from pydantic import ConfigDict, Json


class SharedSettings(BaseSsmParameters):
    """Shared Settings."""

    model_config = ConfigDict(arbitrary_types_allowed=True)

    __ssm_namespace__ = "/algo/shared"

    enappsys: Json[EnappsysSettings]

settings = SharedSettings.from_ssm()

delivery_period_start = dt.datetime(2026, 1, 1, tzinfo=ZoneInfo("Europe/Berlin"))
delivery_period_end = dt.datetime(2026, 9, 1, tzinfo=ZoneInfo("Europe/Berlin"))

enappsys_interface = EnappsysInterface(
    user=settings.enappsys.user,
    password=settings.enappsys.password,
    local_zoneinfo=ZoneInfo("Europe/Berlin"),
    parameter=EnappsysParameter(),
)
enappsys_data = enappsys_interface.load_data(
    delivery_period_start=delivery_period_start,
    delivery_period_end=delivery_period_end,
    data_ids=[EnappsysDataId.RESIDUAL_LOAD_FORECAST]
)
print(enappsys_data)
```

```python
enappsys_data.write_parquet("enappsys_data.parquet")
```
